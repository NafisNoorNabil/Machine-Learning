```python
# Import libraries
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# 1. Load the Iris dataset
iris = load_iris()
X = iris.data       # 4 features
y = iris.target     # 3 flower classes

# 2. Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Apply PCA to reduce 4 features to 2
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 4. Print the results
print("Original shape:", X.shape)
print("Reduced shape:", X_pca.shape)

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance retained:", sum(pca.explained_variance_ratio_))

# 5. Visualize the results
plt.figure(figsize=(8, 6))

for i, flower in enumerate(iris.target_names):
    plt.scatter(
        X_pca[y == i, 0],
        X_pca[y == i, 1],
        label=flower
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA on Iris Dataset")
plt.legend()
plt.show()
```

#Weights
In PCA, weights tell us how much each original feature contributes to a principal component.

```python
# Display the weights (loadings) of each original feature
print("PCA Weights:\n")

for i, component in enumerate(pca.components_):
    print(f"Principal Component {i + 1}:")

    for feature, weight in zip(iris.feature_names, component):
        print(f"{feature}: {weight:.4f}")

    print()
```

```python
# Apply PCA to reduce 4 features to 3
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)

# Check the new shape
print("Reduced shape:", X_pca.shape)

# Check explained variance
print("Explained variance:", pca.explained_variance_ratio_)
print("Total variance retained:", sum(pca.explained_variance_ratio_))
```

```python
# Import 3D plotting tools
import matplotlib.pyplot as plt

# Create a 3D figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot each flower species
for i, flower in enumerate(iris.target_names):

    ax.scatter(
        X_pca[y == i, 0],  # PC1
        X_pca[y == i, 1],  # PC2
        X_pca[y == i, 2],  # PC3
        label=flower
    )

# Add labels and title
ax.set_xlabel("Principal Component 1")
ax.set_ylabel("Principal Component 2")
ax.set_zlabel("Principal Component 3")

ax.set_title("3D PCA - Iris Dataset")
ax.legend()

plt.show()
```
